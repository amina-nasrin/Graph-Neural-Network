```python
import networkx as nx
import numpy as np
```

```python
def flip(p):
  return np.random.random()<.2
```

```python
def all_edges(nodes):
  for i, u in enumerate(nodes):
    for j, v in enumerate(nodes):
      if i<j:
        yield u, v
```

```python
def create_edge(nodes):
  for edge in all_edges(nodes):
    if flip(edge):
      yield (edge)
```

```python
nodes = range(100)

G = nx.Graph()
G.add_nodes_from(nodes)
G.add_edges_from((create_edge(nodes)))

nx.draw(G, node_color='blue', edge_color='gray', node_size=300, with_labels=True)
```
