```python
import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
```

```python
G = nx.erdos_renyi_graph(n=20,p=.15)
```

```python
plt.figure(figsize=(4, 4))
nx.draw(G, with_labels=True)
```

```python
degree = nx.degree_centrality(G)
```

```python
degree
```

```python
closeness = nx.closeness_centrality(G)
```

```python
closeness
```

```python
eigenvector = nx.eigenvector_centrality(G)
```

```python
eigenvector
```

```python
# plot

x = degree.keys()
y1 = degree.values()
y2 = closeness.values()
y3 = eigenvector.values()

fig, ax1 = plt.subplots(1,3,figsize=(14,3))

ax1[0].stem(x, y1, 'k', label = "Degree Centrality")
ax1[1].stem(x, y2, 'b', label = "Closeness Centrality")
ax1[2].stem(x, y3, 'g', label = "Eigenvector Centrality")



for i in range(3):
  ax1[i].legend()
  ax1[i].set(xlim=(-1, 20),ylim=(0, 0.7))


plt.show()
```
